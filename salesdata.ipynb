## Welcome to Sales Analyser

### Reading file

```python
file = open("sales_data.csv", "r")
lines = file.readlines()
```

### Removing first row

```python
newData = []
for line in lines[1:]:
    field = line.strip().split(",")
    field[5] = int(field[5])
    field[6] = int(field[6])
    newData.append(field)
```

### Total revenue from all sales

```python
total_revenue = 0
for row in newData:
    quantity = row[5]
    price = row[6]
    total_revenue = quantity*price

print(f"Total Revenue is Rs. {total_revenue}")
```

### Total oders were placed

```python
total_order = len(newData)
print(f"Total Order : {total_order}")
```

### Total items(Quantity) were sold

```python
total_quantity = 0
for row in newData:
    total_quantity += row[5]

print(f"Total quantity were sold : {total_quantity}")
```

### Total revenue by category

```python
category_revenue = {}

for row in newData:
    category = row[4]
    revenue = row[5]*row[6]

    if category in category_revenue:
        category_revenue[category] += revenue
    else:
        category_revenue[category] = revenue

for cat, rev in category_revenue.items():
    print(f"{cat} : Rs. {rev}")
```

### Total revenue by products

```python
product_revenue = {}

for row in newData:
    product = row[3]
    revenue = row[5]*row[6]

    if product in product_revenue:
        product_revenue[product] += revenue
    else:
        product_revenue[product] = revenue

for pro, rev in product_revenue.items():
    print(f"{pro} : Rs. {rev}")
```

### Product has highest quantity sold

```python
product_quantity = {}

for row in newData:

    product = row[3]
    quantity = row[5]

    if product in product_quantity:
        product_quantity[product] += quantity
    else:
        product_quantity[product] = quantity

highest_sold = None
max_quantity = 0

for pro, qty in product_quantity.items():
    if qty > max_quantity:
        max_quantity = qty
        highest_sold = pro

print(f"Highest sold product : {highest_sold}")
print(f"Quantity sold : {max_quantity}")
```

### Customer spent most money

```python
customer_revenue = {}

for row in newData:
    customer = row[2]
    revenue = row[5]*row[6]

    if customer in customer_revenue:
        customer_revenue[customer] += revenue
    else:
        customer_revenue[customer] = revenue

top_customer = None
max_revenue = 0

for cus, rev in customer_revenue.items():
    if rev > max_revenue:
        max_revenue = rev
        top_customer = cus

print(f"Top Customer : {top_customer}")
print(f"Max Revenue : Rs. {max_revenue}")
```

### Total revenue for each date

```python
daily_revenue = {}
for row in newData:
    date = row[1]
    revenue = row[5]*row[6]

    if date in daily_revenue:
        daily_revenue[date] += revenue
    else:
        daily_revenue[date] = revenue

print("Daily revenue")
# for date, rev in daily_revenue.items():
    # print(f"{date} : Rs. {rev}") # This Loop statement commented due to huge output
```

### Total revenue between two given date

```python
from datetime import datetime

start = input("Enter start date (YYYY-MM-DD)")
end = input("Enter end date (YYYY-MM-DD)")

start_date = datetime.strptime(start, "%Y-%m-%d")
end_date = datetime.strptime(end, "%Y-%m-%d")

total_dateRev = 0

for row in newData:
    order_date = datetime.strptime(row[1], "%Y-%m-%d")

    if start_date <= order_date <= end_date:
        total_dateRev += row[5]*row[6]

print(f"Revenue from {start} to {end} is Rs.. {total_dateRev}")
```

### Electronis sales order in new CSV file

```python
elecFile = open("electronis.csv", "w")
elecFile.write("OrderID, Date, Customer, Product, Category, Quantity, Price\n")

for row in newData:
    if row[4] == "Electronics":
        row_str = ",".join([str(item) for item in row])
        elecFile.write(row_str + "\n")
        
elecFile.close()

print("Electronics.csv file created successfully")
```
